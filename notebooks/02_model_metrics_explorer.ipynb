# Per-class error analysis
Reads python_models/metrics/classical_metrics.json: confusion matrix, per-class F1,
the severe-error list (true class predicted as a *different critical class), and the
candidate sweep history. Mirrors the numbers quoted in the report and the blog.

```python
import json
from pathlib import Path
m = json.load(open('../python_models/metrics/classical_metrics.json'))
t = m['test']
print('winner:', m['winner'])
print('acc', round(t['accuracy'], 4), '| macro-F1', round(t['macro_f1'], 4),
      '| crit recall', round(t['critical_recall'], 4), '| floors met:', m['meets_floors'])
for v in sorted(t['per_class'], key=lambda x: x['f1']):
    flag = ' *critical*' if v['is_critical'] else ''
    print(f"{v['class_name']:26s} F1={v['f1']:.2f} R={v['recall']:.2f}{flag}")
```

```python
# confusion matrix heat map
import matplotlib.pyplot as plt
import numpy as np
names = t['class_names']
cm = t['confusion_matrix']
fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(cm, cmap='Blues')
ax.set_xticks(range(len(names))); ax.set_xticklabels(names, rotation=45, ha='right', fontsize=8)
ax.set_yticks(range(len(names))); ax.set_yticklabels(names, fontsize=8)
for i in range(len(names)):
    for j in range(len(names)):
        ax.text(j, i, cm[i][j], ha='center', va='center', fontsize=8,
                color='white' if cm[i][j] > max(max(r) for r in cm)/2 else 'black')
ax.set_xlabel('predicted'); ax.set_ylabel('true'); ax.set_title('Test confusion matrix')
plt.tight_layout(); plt.show()
print('severe errors (critical->different critical):', len(t['severe_errors']))
```
